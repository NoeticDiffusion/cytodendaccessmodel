```python
from __future__ import annotations

import json
import sys
from pathlib import Path


def detect_repo_root(start: Path | None = None) -> Path:
    start = start or Path.cwd()
    for candidate in (start, *start.parents):
        if (candidate / "README.md").exists() and (candidate / "article").exists():
            return candidate
    return start


REPO_ROOT = detect_repo_root()
SRC_ROOT = REPO_ROOT / "src"
if str(SRC_ROOT) not in sys.path:
    sys.path.insert(0, str(SRC_ROOT))

ARTICLE_DIR = REPO_ROOT / "article" / "Slow Branch-Level Accessibility as a Structural Constraint on Memory Linking"
RESULTS_DIR = REPO_ROOT / "results"


def rel(path: Path) -> str:
    try:
        return str(path.relative_to(REPO_ROOT))
    except ValueError:
        return str(path)


def status(path: Path) -> str:
    return "present" if path.exists() else "missing"


print(f"Repository root: {REPO_ROOT}")
print(f"Article root: {ARTICLE_DIR}")
print(f"Results root: {RESULTS_DIR}")
```
